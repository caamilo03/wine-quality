### <span style="color:#2F749F;"><strong>Ejercicio 1: Cálculo de medidas de tendencia central</strong></span>

Estos fueron los datos utilizados:
   - seed (066047)
   - Media teórica: `(base % 10) + 2`  
   - Desviación estándar: `2`  
   - Tamaño de muestra: `50`  

```python
# Simulación - Ejercicio 1: Medidas de tendencia central

cedula = int(cedula)
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from scipy import stats

np.random.seed(cedula)
normal = np.random.normal(loc=(cedula % 10) + 2, scale=2, size=50)
normal = np.where(normal < 0, 0, normal)  # reemplazar negativos por 0

media   = np.mean(normal)
mediana = np.median(normal)
moda    = float(stats.mode(np.round(normal, 1))[0])

fig, ax = plt.subplots(figsize=(10, 5))

sns.histplot(normal, kde=True, bins=12, color='steelblue', alpha=0.5, ax=ax, label='Distribución')

ax.axvline(media,   color='blue',   linestyle='--', linewidth=2, label=f'Media: {media:.2f}')
ax.axvline(mediana, color='green',  linestyle='-.',  linewidth=2, label=f'Mediana: {mediana:.2f}')
ax.axvline(moda,    color='red',    linestyle=':',   linewidth=2, label=f'Moda: {moda:.2f}')

ax.set_title('Distribución Normal - Medidas de Tendencia Central', fontsize=14)
ax.set_xlabel('Ingresos simulados')
ax.set_ylabel('Frecuencia')
ax.legend()
plt.tight_layout()
plt.show()

print(f"Media:   {media:.4f}")
print(f"Mediana: {mediana:.4f}")
print(f"Moda:    {moda:.4f}")
```

### <span style="color:#2F749F;"><strong>Ejercicio 2: Cálculo de medidas de dispersión</strong></span>

A partir de la muestra generada en el Ejercicio 1, analiza qué tan dispersos están los datos respecto a su media. Calcula:
1. Rango
2. Varianza
3. Desviación estándar
4. Coeficiente de variación
5. Genera una gráfica con los resultados obtenidos (ver imagen de ejemplo)

```python
# Cálculo de medidas de dispersión
media      = np.mean(normal)
minimo     = np.min(normal)
maximo     = np.max(normal)
rango      = maximo - minimo
varianza   = np.var(normal, ddof=1)
desviacion = np.std(normal, ddof=1)
coef_var   = desviacion / media



# ── Gráfica ──────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 5))

sns.histplot(normal, kde=True, bins=12, color='steelblue', alpha=0.5, ax=ax)
ax.axvline(media,              color='blue',   linestyle='--', linewidth=2,   label=f'Media: {media:.2f}')
ax.axvline(media - desviacion, color='orange', linestyle=':',  linewidth=2,   label=f'Media − DE: {media - desviacion:.2f}')
ax.axvline(media + desviacion, color='orange', linestyle=':',  linewidth=2,   label=f'Media + DE: {media + desviacion:.2f}')
ax.axvline(minimo,             color='red',    linestyle='-.', linewidth=1.5, label=f'Mín: {minimo:.2f}')
ax.axvline(maximo,             color='green',  linestyle='-.', linewidth=1.5, label=f'Máx: {maximo:.2f}')
ax.axvspan(media - desviacion, media + desviacion, alpha=0.10, color='orange')
ax.set_title('Distribución con medidas de dispersión', fontsize=13)
ax.set_xlabel('Valores')
ax.set_ylabel('Frecuencia')
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()


print(f"Media:               {media:.4f}")
print(f"Mínimo:              {minimo:.4f}")
print(f"Máximo:              {maximo:.4f}")
print(f"Rango:               {rango:.4f}")
print(f"Varianza:            {varianza:.4f}")
print(f"Desviación estándar: {desviacion:.4f}")
print(f"Coef. de variación:  {coef_var:.4f}  ({coef_var*100:.2f}%)")
```

### <span style="color:#2F749F;"><strong>Ejercicio 3: Cálculo de medidas de distribución</strong></span>

A partir de la muestra generada en el Ejercicio 1, analiza la distribución de los datos. Calcula:
1. Asimetría
2. Curtosis
3. Realiza la comparación con la distribución normal (ver la imagen de ejemplo)

```python
# Cálculo de medidas de asimetría y curtosis
from scipy.stats import skew, kurtosis, norm

asimetria      = skew(normal)
curtosis_valor = kurtosis(normal, fisher=True)  # Fisher=True: curtosis centrada en 0

# ── Gráfica ──────────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(10, 5))

# Histograma + KDE de los datos
sns.histplot(normal, kde=True, bins=12, color='steelblue', alpha=0.5, ax=ax, label='Datos simulados')

# Curva normal teórica para comparar
x = np.linspace(normal.min(), normal.max(), 200)
y = norm.pdf(x, loc=media, scale=desviacion) * len(normal) * (normal.max() - normal.min()) / 12
ax.plot(x, y, color='black', linestyle='--', linewidth=1.5, label='Normal teórica')

# Línea de la media
ax.axvline(media, color='blue', linestyle='--', linewidth=2, label=f'Media: {media:.2f}')

# Anotaciones de asimetría y curtosis en el gráfico
ax.text(0.97, 0.93, f'Asimetría:  {asimetria:.4f}', transform=ax.transAxes,
        ha='right', fontsize=10, color='darkred',
        bbox=dict(boxstyle='round,pad=0.3', facecolor='lightyellow', edgecolor='darkred'))
ax.text(0.97, 0.82, f'Curtosis:   {curtosis_valor:.4f}', transform=ax.transAxes,
        ha='right', fontsize=10, color='purple',
        bbox=dict(boxstyle='round,pad=0.3', facecolor='lavender', edgecolor='purple'))

ax.set_title('Distribución Normal – Asimetría y Curtosis', fontsize=13)
ax.set_xlabel('Valores')
ax.set_ylabel('Frecuencia')
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()

print(f"Asimetría:  {asimetria:.4f}  ({'positiva (cola derecha)' if asimetria > 0 else 'negativa (cola izquierda)' if asimetria < 0 else 'simétrica'})")
print(f"Curtosis:   {curtosis_valor:.4f}  ({'leptocúrtica' if curtosis_valor > 0 else 'platicúrtica' if curtosis_valor < 0 else 'mesocúrtica'})")
```

### <span style="color:#2F749F;"><strong>Ejercicio 5: Análisis de correlación</strong></span>

En esta segunda parte, genera **dos variables** a partir de tú número de cédula y realiza un análisis **bivariado** para calcular la relación entre ellas.

1. Use los **6 últimos dígitos** de su cédula como base (similar al Ejercicio 1).  
2. Genere dos variables aleatorias (ejemplo: `ingresos` y `gastos`) con distribución normal, usando esa base como **semilla reproducible**.  
3. Calcule los coeficientes de correlación de **Pearson** y **Spearman**.
4. Grafique las variables analizadas (ver la imagen de ejemplo)
5. Interprete los resultados.  

```python
from scipy import stats

temp_diaria = np.random.normal(loc=(cedula % 10) + 2, scale=6, size=30)
ventas_helados = np.random.normal(loc=(cedula % 8) + 2, scale=3, size=30)

# Correlación de Pearson
r_pearson, p_pearson = stats.pearsonr(temp_diaria, ventas_helados)

# Correlación de Spearman
r_spearman, p_spearman = stats.spearmanr(temp_diaria, ventas_helados)

print("=" * 55)
print("         CORRELACIONES BIVARIADAS")
print("=" * 55)
print(f"  Pearson:  r = {r_pearson:+.4f}  |  p-valor = {p_pearson:.4f}")
print(f"  Spearman: ρ = {r_spearman:+.4f}  |  p-valor = {p_spearman:.4f}")
print("=" * 55)

def interpret_r(r):
    a = abs(r)
    if a >= 0.9:
        fuerza = "muy fuerte"
    elif a >= 0.7:
        fuerza = "fuerte"
    elif a >= 0.5:
        fuerza = "moderada"
    elif a >= 0.3:
        fuerza = "débil"
    else:
        fuerza = "muy débil o nula"
    direccion = "positiva" if r > 0 else "negativa" if r < 0 else "nula"
    return f"{direccion}, {fuerza}"

print(f"\n  Pearson  → correlación {interpret_r(r_pearson)}")
print(f"  Spearman → correlación {interpret_r(r_spearman)}")

# ── Gráfica 1: Correlación de Pearson (datos originales) ─────────────────
fig1, ax1 = plt.subplots(figsize=(8, 5))

ax1.scatter(temp_diaria, ventas_helados,
            color='steelblue', edgecolors='white', s=70, zorder=3,
            label='Observaciones')

m_p, b_p = np.polyfit(temp_diaria, ventas_helados, 1)
x_line = np.linspace(temp_diaria.min(), temp_diaria.max(), 200)
ax1.plot(x_line, m_p * x_line + b_p, color='tomato', linewidth=2,
         label=f'Regresión lineal  (y = {m_p:.2f}x + {b_p:.2f})')

ax1.text(0.03, 0.97,
         f"Pearson  r = {r_pearson:+.4f}  (p = {p_pearson:.3f})",
         transform=ax1.transAxes, fontsize=9, verticalalignment='top',
         bbox=dict(boxstyle='round,pad=0.4', facecolor='lightyellow',
                   edgecolor='gray', alpha=0.85))

ax1.set_title('Correlación de Pearson\nTemperatura diaria vs Ventas de helados',
              fontsize=12, fontweight='bold', pad=10)
ax1.set_xlabel('Temperatura diaria (°C)', fontsize=11)
ax1.set_ylabel('Ventas de helados (unid.)', fontsize=11)
ax1.legend(fontsize=9)
ax1.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()

# ── Gráfica 2: Correlación de Spearman (datos en rangos) ─────────────────
rang_temp = stats.rankdata(temp_diaria)
rang_vtas = stats.rankdata(ventas_helados)

fig2, ax2 = plt.subplots(figsize=(8, 5))

ax2.scatter(rang_temp, rang_vtas,
            color='mediumseagreen', edgecolors='white', s=70, zorder=3,
            label='Observaciones (rangos)')

m_s, b_s = np.polyfit(rang_temp, rang_vtas, 1)
x_rang = np.linspace(rang_temp.min(), rang_temp.max(), 200)
ax2.plot(x_rang, m_s * x_rang + b_s, color='darkorange', linewidth=2,
         label=f'Regresión sobre rangos  (y = {m_s:.2f}x + {b_s:.2f})')

ax2.text(0.03, 0.97,
         f"Spearman  ρ = {r_spearman:+.4f}  (p = {p_spearman:.3f})",
         transform=ax2.transAxes, fontsize=9, verticalalignment='top',
         bbox=dict(boxstyle='round,pad=0.4', facecolor='honeydew',
                   edgecolor='gray', alpha=0.85))

ax2.set_title('Correlación de Spearman\nTemperatura diaria vs Ventas de helados (rangos)',
              fontsize=12, fontweight='bold', pad=10)
ax2.set_xlabel('Rango — Temperatura diaria', fontsize=11)
ax2.set_ylabel('Rango — Ventas de helados', fontsize=11)
ax2.legend(fontsize=9)
ax2.grid(True, linestyle='--', alpha=0.4)
plt.tight_layout()
plt.show()
```

### <span style="color:#2F749F;"><strong>Ejercicio 6: Chi-Cuadrado</strong></span>

En esta parte, analiza si existe **asociación estadística** entre dos variables categóricas simuladas a partir de tu número de cédula.

1. Usa los **6 últimos dígitos** de tu cédula como base para generar una semilla reproducible.
2. Simula una muestra de **60 personas** con dos variables categóricas:
   - `Nivel educativo`: Básico, Medio, Superior.
   - `Acceso a internet`: Sí, No.
3. Crea una **tabla de contingencia** entre ambas variables.
4. Aplica la **prueba de Chi cuadrado de independencia**.
5. Visualiza la distribución conjunta con un mapa de calor (ver la imagen de con la información de ejemplo).
6. Interpreta el resultado.

image.png

```python
import numpy as np
import pandas as pd
from IPython.display import display

# Fijar semilla
np.random.seed(cedula)

# Generar datos aleatorios
n = 60

nivel_educativo = np.random.choice(["Básico", "Medio", "Superior"], size=n)
acceso_internet = np.random.choice(["Sí", "No"], size=n)

# Crear DataFrame
df = pd.DataFrame({
    "Nivel Educativo": nivel_educativo,
    "Acceso a Internet": acceso_internet
})

# Tabla de contingencia con totales
tabla = pd.crosstab(
    df["Nivel Educativo"],
    df["Acceso a Internet"],
    margins=True,
    margins_name="Total"
)

# ── Tabla estilizada ──────────────────────────────────────────────────────
# Separar datos internos de la fila/columna Total
data_cells  = tabla.iloc[:-1, :-1]   # sin fila Total ni columna Total
total_row   = tabla.iloc[-1:, :]      # fila Total completa
total_col   = tabla.iloc[:, -1:]      # columna Total completa

def style_tabla(df_tabla):
    # Paleta de colores
    color_header  = "#2F749F"      # azul encabezado
    color_total   = "#D6E4F0"      # azul claro para totales
    color_low     = "#EAF4FB"      # celda baja
    color_high    = "#2874A6"      # celda alta
    text_total    = "#1A5276"

    styled = (
        df_tabla.style
        .background_gradient(
            cmap="Blues",
            subset=pd.IndexSlice[df_tabla.index[:-1], df_tabla.columns[:-1]],
            vmin=0
        )
        .set_properties(
            subset=pd.IndexSlice[df_tabla.index[-1], :],   # fila Total
            **{"background-color": color_total,
               "color": text_total,
               "font-weight": "bold"}
        )
        .set_properties(
            subset=pd.IndexSlice[:, df_tabla.columns[-1]],  # columna Total
            **{"background-color": color_total,
               "color": text_total,
               "font-weight": "bold"}
        )
        .set_properties(**{
            "text-align": "center",
            "font-size": "13px",
            "border": "1px solid #AED6F1",
            "padding": "8px 14px"
        })
        .set_table_styles([
            # Encabezado de columnas
            {"selector": "thead th",
             "props": [("background-color", color_header),
                       ("color", "white"),
                       ("font-size", "13px"),
                       ("font-weight", "bold"),
                       ("text-align", "center"),
                       ("padding", "8px 14px"),
                       ("border", "1px solid #2471A3")]},
            # Índice (columna izquierda)
            {"selector": "tbody th",
             "props": [("background-color", color_header),
                       ("color", "white"),
                       ("font-weight", "bold"),
                       ("text-align", "center"),
                       ("padding", "8px 14px"),
                       ("border", "1px solid #2471A3")]},
            # Tabla general
            {"selector": "table",
             "props": [("border-collapse", "collapse"),
                       ("margin", "12px auto"),
                       ("box-shadow", "2px 2px 6px rgba(0,0,0,0.15)")]},
            # Caption
            {"selector": "caption",
             "props": [("caption-side", "top"),
                       ("font-size", "15px"),
                       ("font-weight", "bold"),
                       ("color", color_header),
                       ("padding-bottom", "8px")]}
        ])
        .set_caption("Tabla de contingencia — Nivel Educativo × Acceso a Internet")
        .format("{:d}")
    )
    return styled

display(style_tabla(tabla))
```

```python
from scipy.stats import chi2_contingency

# ── Prueba Chi-Cuadrado de independencia ─────────────────────────────────
tabla_vals = tabla.iloc[:-1, :-1]   # sin fila/columna Total
chi2, p_valor, gl, esperados = chi2_contingency(tabla_vals)

print("=" * 52)
print("     PRUEBA CHI-CUADRADO DE INDEPENDENCIA")
print("=" * 52)
print(f"  Estadístico χ²  : {chi2:.4f}")
print(f"  Grados de libertad: {gl}")
print(f"  p-valor          : {p_valor:.4f}")
print("=" * 52)
alpha = 0.05
if p_valor < alpha:
    print(f"\n  → p = {p_valor:.4f} < α = {alpha}")
    print("  → Se RECHAZA H₀: existe asociación significativa")
    print("    entre el nivel educativo y el acceso a internet.")
else:
    print(f"\n  → p = {p_valor:.4f} ≥ α = {alpha}")
    print("  → No se rechaza H₀: NO hay evidencia suficiente")
    print("    de asociación entre las variables.")

# ── Mapa de calor ─────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(7, 5))

obs = tabla_vals.copy()
sns.heatmap(
    obs,
    ax=ax,
    annot=True, fmt="d",
    cmap="Blues",
    linewidths=0.6, linecolor="white",
    cbar_kws={"label": "Frecuencia observada"},
    annot_kws={"size": 14, "weight": "bold"}
)
ax.set_xlabel("Acceso a Internet", fontsize=11)
ax.set_ylabel("Nivel Educativo", fontsize=11)
ax.tick_params(axis="x", rotation=0)
ax.tick_params(axis="y", rotation=0)

sig = "✓ Asociación significativa" if p_valor < alpha else "✗ Sin asociación significativa"
fig.suptitle(
    f"Frecuencias Observadas\nχ² = {chi2:.3f},  p = {p_valor:.4f}  —  {sig}",
    fontsize=12, fontweight="bold"
)
plt.tight_layout()
plt.show()
```

```python
# Librerías
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url_red = 'https://raw.githubusercontent.com/caamilo03/wine-quality/main/datos/winequality-red.csv'
url_white = 'https://raw.githubusercontent.com/caamilo03/wine-quality/main/datos/winequality-white.csv'

df_red = pd.read_csv(url_red, sep=';')
df_white = pd.read_csv(url_white, sep=';')

df_red.head()
```

```python
# Importar librerías para PCA
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Añadir columna de tipo de vino y combinar
df_red_pca   = df_red.copy()
df_white_pca = df_white.copy()
df_red_pca['tipo']   = 'Tinto'
df_white_pca['tipo'] = 'Blanco'

df = pd.concat([df_red_pca, df_white_pca], ignore_index=True)

print(f"Dataset combinado: {df.shape[0]} vinos")
print(f"  - Vinos tintos:  {df_red.shape[0]}")
print(f"  - Vinos blancos: {df_white.shape[0]}")
print(f"  - Variables fisicoquimicas: {df.shape[1] - 2}  (excl. 'quality' y 'tipo')")
df.head()
```

<div style="background-color:#e6f2ff; padding:20px; border-radius:10px;">
<b></b> Seleccionar variables, estandarizar y explorar correlaciones
</div>

```python
# Variables fisicoquimicas (excluir quality y tipo)
features = [col for col in df.columns if col not in ['quality', 'tipo']]
X = df[features]

# Estandarizar (media=0, desviacion=1) — requisito del PCA
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled_df = pd.DataFrame(X_scaled, columns=features)

print(f"Numero de variables fisicoquimicas: {len(features)}")
print(f"Variables: {features}\n")
print("Estadisticas post-estandarizacion (deben ser ~0 y ~1):")
print(X_scaled_df.describe().loc[['mean', 'std']].round(3).to_string())

# Matriz de correlacion
plt.figure(figsize=(12, 9))
corr_matrix = X_scaled_df.corr()
mask = np.triu(np.ones_like(corr_matrix, dtype=bool))  # solo triangulo inferior
sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm',
            center=0, linewidths=0.5, annot_kws={'size': 8},
            mask=mask)
plt.title('Matriz de correlacion - Variables fisicoquimicas del vino',
          fontsize=14, fontweight='bold', backgroundcolor='#fff9db')
plt.tight_layout()
plt.show()
```

<div style="background-color:#e6f2ff; padding:20px; border-radius:10px;">
 Aplicar PCA y analizar la varianza explicada
</div>

```python
# Aplicar PCA completo (todos los componentes)
pca = PCA()
pca.fit(X_scaled)

var_exp  = pca.explained_variance_ratio_
var_acum = np.cumsum(var_exp)

# Criterios para elegir numero de componentes
n_kaiser = int(np.sum(pca.explained_variance_ > 1))
n_80 = int(np.where(var_acum >= 0.80)[0][0] + 1)

print(f"Componentes con autovalor > 1 (Criterio Kaiser): {n_kaiser}")
print(f"Componentes para explicar >= 80% de varianza:    {n_80}\n")

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Scree Plot ---
axes[0].bar(range(1, len(var_exp)+1), var_exp*100,
            color='steelblue', alpha=0.7, label='Varianza individual')
axes[0].plot(range(1, len(var_exp)+1), var_exp*100,
             'o-', color='steelblue', linewidth=2)
axes[0].axhline(y=100/len(features), color='red', linestyle='--', linewidth=1.5,
                label=f'Criterio Kaiser ({100/len(features):.1f}%)')
axes[0].set_xlabel('Componente principal')
axes[0].set_ylabel('Varianza explicada (%)')
axes[0].set_title('Grafico de sedimentacion (Scree Plot)',
                   fontsize=12, fontweight='bold', backgroundcolor='#fff9db')
axes[0].set_xticks(range(1, len(var_exp)+1))
axes[0].legend()

# --- Varianza acumulada ---
axes[1].plot(range(1, len(var_acum)+1), var_acum*100,
             'o-', color='darkorange', linewidth=2)
axes[1].axhline(y=80, color='gray', linestyle='--', linewidth=1.5, label='80% varianza')
axes[1].fill_between(range(1, len(var_acum)+1), var_acum*100,
                     alpha=0.2, color='darkorange')
axes[1].set_xlabel('Numero de componentes')
axes[1].set_ylabel('Varianza acumulada (%)')
axes[1].set_title('Varianza acumulada explicada',
                   fontsize=12, fontweight='bold', backgroundcolor='#fff9db')
axes[1].set_xticks(range(1, len(var_acum)+1))
axes[1].legend()

plt.tight_layout()
plt.show()

# Tabla resumen
df_var = pd.DataFrame({
    'Componente':         [f'PC{i+1}' for i in range(len(var_exp))],
    'Autovalor':          pca.explained_variance_.round(3),
    'Var. Explicada (%)': (var_exp*100).round(2),
    'Var. Acumulada (%)': (var_acum*100).round(2)
})
print("Tabla de varianza explicada:")
print(df_var.to_string(index=False))
```

<div style="background-color:#f2f2f2; padding:10px; border-radius:5px;">
  <b>Interpretacion: varianza explicada y numero de componentes</b><br>
  <ul>
    <li>El <strong>scree plot</strong> muestra la varianza que aporta cada componente. El "codo" de la curva indica el numero optimo de componentes a retener.</li>
    <li>El <strong>criterio de Kaiser</strong> sugiere retener los componentes cuyo autovalor supere 1 (explican mas varianza que una variable estandarizada individual).</li>
    <li>La <strong>varianza acumulada</strong> permite determinar cuantos componentes son necesarios para retener un umbral deseado (p.ej., 80% de la variabilidad total).</li>
    <li>Seleccionar menos componentes simplifica el analisis pero reduce el detalle; mas componentes preservan mas informacion pero aumentan la complejidad.</li>
  </ul>
</div>

<div style="background-color:#e6f2ff; padding:20px; border-radius:10px;">
Construir el Biplot (proyeccion de observaciones + vectores de carga)
</div>

```python
# PCA con 2 componentes para visualización
pca2 = PCA(n_components=2)
X_pca2 = pca2.fit_transform(X_scaled)

fig, ax = plt.subplots(figsize=(12, 9))

# Scatter plot coloreado por tipo de vino (muestra aleatoria para legibilidad)
np.random.seed(42)
sample_idx = np.random.choice(len(df), size=min(800, len(df)), replace=False)

colors = {'Tinto': 'crimson', 'Blanco': '#DAA520'}
for tipo, color in colors.items():
    mask = np.array(df['tipo'] == tipo)
    idx  = np.where(mask)[0]
    idx_sample = np.intersect1d(idx, sample_idx)
    ax.scatter(X_pca2[idx_sample, 0], X_pca2[idx_sample, 1],
               c=color, label=tipo, alpha=0.3, s=18, edgecolors='none')

# Vectores de carga (loadings arrows escalados)
loadings_raw = pca2.components_.T
scale_factor = 3.5
for i, feat in enumerate(features):
    ax.arrow(0, 0,
             loadings_raw[i, 0] * scale_factor,
             loadings_raw[i, 1] * scale_factor,
             head_width=0.10, head_length=0.08,
             fc='navy', ec='navy', linewidth=1.5, zorder=5)
    ax.text(loadings_raw[i, 0] * scale_factor * 1.18,
            loadings_raw[i, 1] * scale_factor * 1.18,
            feat, fontsize=9, ha='center', va='center',
            color='navy', fontweight='bold', zorder=6)

ax.axhline(0, color='gray', linestyle='--', alpha=0.4, linewidth=0.8)
ax.axvline(0, color='gray', linestyle='--', alpha=0.4, linewidth=0.8)
ax.set_xlabel(f'PC1  ({pca2.explained_variance_ratio_[0]*100:.1f}% varianza)', fontsize=12)
ax.set_ylabel(f'PC2  ({pca2.explained_variance_ratio_[1]*100:.1f}% varianza)', fontsize=12)
ax.set_title('Biplot PCA - Vinos tinto y blanco\n(puntos = observaciones, flechas = contribucion de cada variable)',
             fontsize=13, fontweight='bold', backgroundcolor='#fff9db')
ax.legend(title='Tipo de vino', fontsize=11, title_fontsize=11)
plt.tight_layout()
plt.show()
```

<div style="background-color:#e6f2ff; padding:20px; border-radius:10px;">
<b>Paso 4:</b> Analizar las cargas factoriales (loadings)
</div>

```python
# PCA con 3 componentes para analizar cargas
pca3 = PCA(n_components=3)
pca3.fit(X_scaled)

# Cargas factoriales (correlación variable-componente)
loadings_df = pd.DataFrame(
    pca3.components_.T,
    columns=['PC1', 'PC2', 'PC3'],
    index=features
)

# Ordenar por valor absoluto de PC1 (mayor contribución primero)
loadings_df = loadings_df.reindex(
    loadings_df['PC1'].abs().sort_values(ascending=False).index
)

# 1. Heatmap de cargas
plt.figure(figsize=(8, 6))
sns.heatmap(loadings_df, annot=True, fmt='.3f', cmap='coolwarm',
            center=0, linewidths=0.5, annot_kws={'size': 10},
            vmin=-1, vmax=1)
plt.title('Cargas factoriales (Loadings)\nPC1, PC2 y PC3',
          fontsize=12, fontweight='bold', backgroundcolor='#fff9db')
plt.ylabel('Variable fisicoquímica')
plt.xlabel('Componente principal')
plt.tight_layout()
plt.show()

# 2. Gráfico de barras de contribución conjunto para PC1 y PC2
plt.figure(figsize=(10, 5))
x = np.arange(len(features))
width = 0.35

plt.bar(x - width/2, pca3.components_[0], width, label='PC1', color='lightblue')
plt.bar(x + width/2, pca3.components_[1], width, label='PC2', color='steelblue')

plt.axhline(0, color='black', linewidth=0.8)
plt.xticks(x, features, rotation=45, ha='right', fontsize=9)
plt.ylabel('Carga factorial')
plt.title('Contribución de variables a PC1 y PC2',
          fontsize=12, fontweight='bold', backgroundcolor='#fff9db')
plt.legend()
plt.tight_layout()
plt.show()

# 3. Gráfica de proporción de varianza explicada (Estilo imagen)
# Nota: usamos pca completo si existe, de lo contrario usamos pca3
pca_full = pca if 'pca' in locals() else pca3
var_exp = pca_full.explained_variance_ratio_
var_acum = np.cumsum(var_exp)

x_pos = np.arange(len(var_exp))
labels = [f'PC{i+1}' for i in range(len(var_exp))]

fig, ax = plt.subplots(figsize=(10, 6))

# Barras para PVE individual
bars = ax.bar(x_pos, var_exp, color='skyblue', label='PVE individual', alpha=0.9)

# Línea PVE acumulada
ax.plot(x_pos, var_acum, marker='o', markersize=5, color='navy', label='PVE acumulada')

# Umbrales
ax.axhline(y=0.7, color='red', linestyle='--', label='Umbral 70%')
ax.axhline(y=0.9, color='green', linestyle='--', label='Umbral 90%')

ax.set_xticks(x_pos)
ax.set_xticklabels(labels)
ax.set_xlabel('Componentes principales')
ax.set_ylabel('Proporción de varianza')
ax.set_title('Proporción de varianza explicada', 
             fontsize=14, fontweight='bold', backgroundcolor='#fff9db', pad=15)

# Anotaciones sobre las barras
for i, b in enumerate(bars):
    height = b.get_height()
    ax.text(b.get_x() + b.get_width()/2., height + 0.01,
            f'{height:.2f}', ha='center', va='bottom', fontsize=9)

ax.grid(True, axis='y', linestyle='--', alpha=0.5)
# Reordenar leyendas
handles, chart_labels = ax.get_legend_handles_labels()
ax.legend(handles, chart_labels, loc='lower right', framealpha=1)
plt.tight_layout()
plt.show()

# 4. Gráfica de autovalores
autovalores = pca_full.explained_variance_

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(autovalores) + 1), autovalores, 'o-', color='purple', linewidth=2)
plt.axhline(1, color='red', linestyle='--', label='Criterio de Kaiser (Autovalor=1)')
plt.xlabel('Componente principal')
plt.ylabel('Autovalor')
plt.title('Autovalores por Componente Principal',
          fontsize=12, fontweight='bold', backgroundcolor='#fff9db')
plt.xticks(range(1, len(autovalores) + 1))
plt.legend()
plt.tight_layout()
plt.show()

print("\nCargas factoriales (primeras 3 componentes):")
print(loadings_df.round(3).to_string())
```
